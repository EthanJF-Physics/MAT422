# Linear Spaces

## Linear Combinations

**Definition 1** *(Linear subspace)*. Let $U \subseteq V$ be a subset. $U$ is a linear subspace if it is closed under vector addition and scalar multiplication. That is, for all $u_1, u_2 \in U$ and $\alpha \in \mathbb{R}$, it holds that
$$u_1 + u_2 \in U \quad \text{and} \quad \alpha u_1 \in U$$
In particular, $0$ is always in a linear subspace. As we can see below, the span of a set of vectors is a linear subspace.

## Linear Independence

**Definition 2** *(Linear independence)*. A list of vectors $u_1, \dots, u_m$ is linearly independent if none of them can be written as a linear combination of the others—that is,
$$\forall i, \quad u_i \notin \operatorname{span}(\{u_j \mid j \neq i\})$$
A list of vectors is called linearly dependent if it is not linearly independent.

**Lemma 1** *(Characterization of linear independence)*. The vectors $u_1, \dots, u_m$ are linearly independent if and only if
$$\sum_{j=1}^{m} \alpha_j u_j = 0 \implies \alpha_j = 0 \quad \forall j$$
Equivalently, $u_1, \dots, u_m$ are linearly dependent if and only if there exist $\alpha_j$'s, not all zero, such that
$$\sum_{j=1}^{m} \alpha_j u_j = 0$$

**Theorem 1** *(Dimension theorem)*. Let $U$ be a linear subspace of $V$. Any basis of $U$ always has the same number of elements. All bases of $U$ have the same length, that is, the same number of elements. We call this number the dimension of $U$ and denote it $\dim(U)$.

**Lemma 2** *(Characterization of linearly dependent sets)*. Let $u_1, \dots, u_m$ be a linearly dependent list of vectors with a linearly independent subset $u_1, \dots, u_k$, where $k < m$. Then there is an $i > k$ such that:
1. $u_i \in \operatorname{span}(u_1, \dots, u_{i-1})$;
2. $\operatorname{span}(\{u_j \mid j \in \{1, \dots, m\}\}) = \operatorname{span}(\{u_j \mid j \in \{1, \dots, m\}, j \neq i\})$

## Orthonormal Bases

**Definition 3** *(Norm and inner product)*. For any vectors $u, v \in V$, the standard inner product and norm are defined as
$$\langle u, v \rangle = u \cdot v = \sum_{i=1}^{n} u_i v_i \quad \text{and} \quad \|u\| = \sqrt{\sum_{i=1}^{n} u_i^2}$$

**Definition 4** *(Orthonormal list)*. A list of vectors $u_1, \dots, u_m$ is orthonormal if the $u_i$'s are pairwise orthogonal and each has norm 1, that is, for all $i$ and all $j \neq i$,
$$\langle u_i, u_j \rangle = 0 \quad \text{and} \quad \|u_i\| = 1$$

**Lemma 3** *(Cauchy–Schwarz)*. For any $u, v \in V$, the inner product is bounded by the product of their norms:
$$|\langle u, v \rangle| \le \|u\|\|v\|$$

## Best Approximation Theorem

Many optimization applications can be converted to the following best approximation problem. We have a linear subspace $U \subseteq V$ and a vector $v \notin U$. We want to find the vector $v^* \in U$ that is closest to $v$ in the norm, that is, we want to solve
$$\min_{v^* \in U} \|v^* - v\|$$

**Definition 5** *(Orthogonal projection)*. Let $U \subseteq V$ be a linear subspace with orthonormal basis $q_1, \dots, q_m$. The orthogonal projection of $v \in V$ on $U$ is defined as
$$P_U v = \sum_{j=1}^{m} \langle v, q_j \rangle q_j$$

**Theorem 2** *(Best approximation theorem)*. Let $U \subseteq V$ be a linear subspace with orthonormal basis $q_1, \dots, q_m$ and let $v \in V$. For any $u \in U$,
$$\|v - P_U v\| \le \|v - u\|$$
Furthermore, if $u \in U$ and the inequality above is an equality, then $u = P_U v$.

**Lemma 4** *(Orthogonal decomposition)*. Let $U \subseteq V$ be a linear subspace with orthonormal basis $q_1, \dots, q_m$ and let $v \in V$. For any $u \in U$,
$$\langle v - P_U v, u \rangle = 0$$
In particular, $v$ can be decomposed as $(v - P_U v) + P_U v$ where the two terms are orthogonal.

We can test our best approximation by constructing a matrix $Q$ with columns $q_1, \dots, q_m$ and checking that:
$$Q^T Q = \begin{pmatrix}
\langle q_1, q_1 \rangle & \cdots & \langle q_1, q_m \rangle \\
\langle q_2, q_1 \rangle & \cdots & \langle q_2, q_m \rangle \\
\vdots & \ddots & \vdots \\
\langle q_m, q_1 \rangle & \cdots & \langle q_m, q_m \rangle
\end{pmatrix} = I_{m \times m}$$
where $I_{m \times m}$ denotes the $m \times m$ identity matrix.


```python
# 1.2.2 Orthogonality
# Orthonormal Bases
def inner_product(u, v):
    return np.dot(u, v)
def norm(b):
    return np.sqrt(inner_product(b, b))
# Show if orthogoal/orthonormal vectors
def check_ortho(u, v):
    is_gonal = np.isclose(inner_product(u, v), 0.0)
    is_normal = is_gonal and np.isclose(norm(u), 1.0) and np.isclose(norm(v), 1.0)
    return print("Orthgonal?", is_gonal, "Orthonormal?", is_normal)
# Show Cauchy-Swartz THM
def cauchy_schwarz(u, v):
    lhs = abs(inner_product(u, v))
    rhs = norm(u) * norm(v)
    result = lhs <= rhs or np.isclose(lhs, rhs)
    print("Cauchy-Schwarz Satisfied?", result)
    return result
# Show Orthogonal Proj and Decomp
def o_decomp(u, v):
    proj_v_on_u = (inner_product(v, u)) / inner_product(u, u) * u
    prep_v_on_u = v - proj_v_on_u
    print("Orthogonal Decompostion's Projection of v on u", proj_v_on_u)
    # verify using best approx thm
    is_orthogonal_residual = np.isclose(inner_product(prep_v_on_u, u), 0.0)
    print(f"  Best Approximation Check (perp ⊥ u): {is_orthogonal_residual}")
    return proj_v_on_u, prep_v_on_u

# Now Example with u1 and u2
print("\nOrthogonality & Orthonormality Check")
check_ortho(u1, u2)
print("\nCauchy-Schwarz Theorem Verification")
cauchy_schwarz(u1, u2)
print("\nOrthogonal Decomposition & Best Approximation")
proj_vector, perp_vector = o_decomp(u1, u2)
print("Orthogonal Decomposition's Perpendicular component:", perp_vector)
```

**Theorem 3** *(Gram–Schmidt)*. Let $a_1, \dots, a_m$ in $\mathbb{R}^n$ be linearly independent. Then there exists an orthonormal basis $q_1, \dots, q_m$ of $\operatorname{span}(a_1, \dots, a_m)$.

## Gram–Schmidt Process

1. **Initialization ($k=0$):**
   - Code: `U[:, 0] = M[:, 0]` and `E[:, 0] = U[:, 0] / np.linalg.norm(U[:, 0])`
   - Takes the first vector $a_1$ and normalizes it to create the first orthonormal basis vector:
     $$q_1 = \frac{a_1}{\|a_1\|}$$
2. **Sequential Orthogonalization ($k = 1, \dots, m-1$):**
   - Code: `u_k = M[:, k].copy()` followed by the inner loop subtracting projections.
   - For each subsequent vector $a_k$ (the $k$-th column), we remove its components along all previously established orthogonal directions $u_1, \dots, u_{k-1}$ (or their normalized versions $q_j$). This isolates the part of $a_k$ that is strictly orthogonal to the subspace spanned by the previous vectors:
     $$u_k = a_k - \sum_{j=1}^{k-1} \langle a_k, q_j \rangle q_j$$
3. **Normalization:**
   - Code: `E[:, k] = U[:, k] / np.linalg.norm(U[:, k])`
   - Scales the newly orthogonalized vector $u_k$ to have a norm of 1, yielding the next orthonormal basis vector $q_k$:
     $$q_k = \frac{u_k}{\|u_k\|}$$
4. **Verification ($Q^T Q = I$):**
   - Code: `np.dot(gs_A.T, gs_A)`
   - Checks that all column vectors $q_i$ are mutually orthogonal ($\langle q_i, q_j \rangle = 0$ for $i \neq j$) and normalized ($\|q_i\|^2 = 1$), confirming that $Q^T Q = I_{m \times m}$ as established in the theory.

```python
# 1.2.3 Gran-Schmidt Process
def gram_schmidt_matrix(Matrix_Input):
    M = Matrix_Input.astype('float64')
    rows, cols = M.shape
    # U holds the orthogonal vectors, E holds the normalized orthonormal vectors
    U = np.zeros((rows, cols))
    E = np.zeros((rows, cols))
    # Process first column
    U[:, 0] = M[:, 0]
    E[:, 0] = U[:, 0] / np.linalg.norm(U[:, 0])
    # Process remaining columns
    for k in range(1, cols):
        u_k = M[:, k].copy()
        # Subtract projections onto all previously established orthogonal vectors
        for j in range(k):
            u_k -= proj(U[:, j], M[:, k])
        U[:, k] = u_k
        E[:, k] = U[:, k] / np.linalg.norm(U[:, k])
    return E
# Calculate the Gram-Schmidt Matrix of A
gs_A = gram_schmidt_matrix(A)
print("Gram-Schmidt Orthonormal Matrix of A:\n", gs_A)
# Verify identity (Q^T * Q = I)
identity_check = np.dot(gs_A.T, gs_A)
print("\nVerification (Identity Matrix Check gs_A^T * gs_A):\n", np.round(identity_check, 6))
```

## Eigenvalues and Eigenvectors

**Definition 6** *(Eigenvalues and eigenvectors)*. Let $A \in \mathbb{R}^{d \times d}$ be a square matrix. Then $\lambda \in \mathbb{R}$ is an eigenvalue of $A$ if there exists a non-zero vector $x \neq 0$ such that
$$Ax = \lambda x$$
The vector $x$ is referred to as an eigenvector.

**Lemma 5** *(Number of eigenvalues)*. Let $A \in \mathbb{R}^{d \times d}$ and let $\lambda_1, \dots, \lambda_m$ be distinct eigenvalues of $A$ with corresponding non-zero eigenvectors $x_1, \dots, x_m$. Then $x_1, \dots, x_m$ are linearly independent. As a result, $m \le d$.

**Theorem 4** *(Orthogonality of eigenvectors)*. If $A$ is symmetric, then any two eigenvectors from different eigenspaces are orthogonal.

**Theorem 5** *(The spectral theorem for symmetric matrices)*. An $n \times n$ symmetric matrix $A$ has the following properties:
* $A$ has $n$ real eigenvalues, counting multiplicities.
* If $\lambda$ is an eigenvalue of $A$ with multiplicity $k$, then the eigenspace for $\lambda$ is $k$-dimensional.
* The eigenspaces are mutually orthogonal, in the sense that eigenvectors corresponding to different eigenvalues are orthogonal.
* $A$ is orthogonally diagonalizable.

**Theorem 6** *(Rayleigh quotient optimization)*. Let $A$ be an $n \times n$ symmetric matrix with an orthogonal diagonalization $A = PDP^{-1}$ (where $P^{-1} = P^T$). The columns of $P$ are orthonormal eigenvectors $v_1, \dots, v_n$ of $A$. Assume that the diagonals of $D$ are arranged so that $\lambda_1 \le \lambda_2 \le \dots \le \lambda_n$. Then
$$\min_{x \neq 0} \frac{x^T A x}{x^T x} = \lambda_1$$
is achieved when $x = v_1$, and
$$\max_{x \neq 0} \frac{x^T A x}{x^T x} = \lambda_n$$
is achieved when $x = v_n$.


```python
# 1.2.4 Eigenvalues and Eigenvectors
# Calculate Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A.astype(float))
print("Eigenvalues of Matrix A:\n", eigenvalues)
print("\nEigenvectors of Matrix A (as columns):\n", eigenvectors)

# Verification of A * v = lambda * v for the first eigenvalue/eigenvector pair
lambda_1 = eigenvalues[0]
v_1 = eigenvectors[:, 0]
left_side = np.dot(A, v_1)
right_side = lambda_1 * v_1
print("\nVerification of A * v = lambda * v for first pair:")
print("A * v_1:\n", left_side)
print("lambda_1 * v_1:\n", right_side)
print(f"Are both sides equal? {np.allclose(left_side, right_side)}")
```
